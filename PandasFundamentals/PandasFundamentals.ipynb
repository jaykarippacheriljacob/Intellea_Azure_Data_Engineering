## 2.3 Selecting by label and position: `loc` and `iloc`

| Syntax | Selects by | Slice end |
|---|---|---|
| `df.loc[row_label]` | Row **label** (index name) | **included**: `loc[0:2]` → rows 0, 1, 2 |
| `df.loc[row_label, "col"]` | Row label + column name | |
| `df.iloc[row_index, col_index]` | **Position** (like NumPy) | excluded: `iloc[0:2]` → rows 0, 1 |

With the default index (0, 1, 2…) labels and positions look the same; the difference shows in slicing and after filtering or sorting.

````python
print(orders.loc[0, "Order_ID"])            # row label 0, column Order_ID
print(orders.iloc[0, 0])                    # row position 0, column position 0
print(len(orders.loc[0:2]), len(orders.iloc[0:2]))   # 3 vs 2 → loc includes the end
````

````python
orders.loc[0:2, ["Order_ID", "Quantity", "Order_Status"]]
````

````python
orders.iloc[0:3, 0:3]          # first 3 rows, first 3 columns
````

## 2.4 Filtering rows

A comparison gives a True/False **mask**; `df[mask]` keeps only the True rows.

| Operator | Meaning |
|---|---|
| `>`, `<`, `>=`, `<=`, `==`, `!=` | Comparisons |
| `&` | AND |
| `\|` | OR |
| `~` | NOT |

Use `&`, `|`, `~` (not `and`, `or`, `not`), and put **each condition in brackets**.

````python
orders[orders["Quantity"]>2].head()
````

### Filter rows and keep some columns

| Way | Code |
|---|---|
| Chained | `df[condition][["a", "b"]]` |
| **`loc` (preferred)** | `df.loc[condition, ["a", "b"]]` (rows, columns in one step) |

`loc` avoids the `SettingWithCopyWarning` if you later change the result.

````python
orders[orders["Quantity"] > 2][["Order_ID", "Customer_ID", "Quantity"]].head()
````

````python
orders.loc[orders["Quantity"] > 2, ["Order_ID", "Customer_ID", "Quantity"]].head()
````

### Multiple conditions

````python
orders[(orders["Quantity"]>2) & (orders["Order_Status"] == "Completed")].head()
````

````python
# OR: Cash on Delivery or UPI
cod_upi = orders[(orders["Payment_Method"] == "Cash on Delivery") | (orders["Payment_Method"] == "UPI")]
print("COD or UPI:", len(cod_upi))

# NOT: every order that was not Completed
print("Not completed:", len(orders[~(orders["Order_Status"] == "Completed")]))
````

## 2.10 Common workflow and mistakes

| Step | Do |
|---|---|
| 1 | Inspect: `head()`, `info()`, `describe()` |
| 2 | Select the columns you need |
| 3 | Filter rows with conditions |
| 4 | Check missing values and duplicates |
| 5 | Clean text, convert data types |
| 6 | Re-check the result (`isna().sum()`, `duplicated().sum()`, `shape`) |

| Mistake | Fix |
|---|---|
| Forgetting brackets in multiple conditions | `(a > 1) & (b == "x")` |
| `and` / `or` instead of `&` / `\|` | Use `&`, `\|`, `~` |
| Filtering before fixing data types | Convert first (e.g. dates) |
| `dropna()` without checking the loss | Compare `shape` before and after |
| `df.dtype` on a DataFrame | `df.dtypes` (plural); `dtype` is for one column |

## 3.2 Combining datasets: `merge()`

```python
df = orders.merge(products, on="Product_ID", how="left")
```

`on` is the **key** column both tables share.

| `how` | Keeps |
|---|---|
| `"left"` | **All** rows of the left table (`orders`) + matching info from the right; no match → `NaN` |
| `"inner"` | Only rows that match in both |
| `"right"` | All rows of the right table |
| `"outer"` | All rows of both |

````python
df = orders.merge(products, on="Product_ID", how="left")
df.head()
````

## 3.3 Computed columns

`df["new"] = ...` creates a column from whole-column arithmetic (no loop).

| Column | Formula |
|---|---|
| `Gross_Sales` | `Quantity × Unit_Price_INR` |
| `Revenue_INR` | `Gross_Sales × (1 − Discount_Pct / 100)` |
| `Total_Cost_INR` | `Quantity × Unit_Cost_INR` |
| `Profit_INR` | `Revenue_INR − Total_Cost_INR` |

````python
df["Gross_Sales"] = df["Quantity"] * df["Unit_Price_INR"]
df[["Quantity", "Unit_Price_INR", "Gross_Sales"]].head()
````

````python
df["Revenue_INR"] = df["Gross_Sales"] * (1 - df["Discount_Pct"]/100)
df[["Quantity", "Unit_Price_INR", "Gross_Sales", "Revenue_INR"]].head()
````

````python
df["Total_Cost_INR"] = df["Quantity"] * df["Unit_Cost_INR"]
df[["Quantity", "Unit_Price_INR", "Gross_Sales", "Revenue_INR", "Total_Cost_INR"]].head()
````

````python
df["Profit_INR"] = df["Revenue_INR"] - df["Total_Cost_INR"]
df[["Quantity", "Unit_Price_INR", "Gross_Sales", "Revenue_INR", "Total_Cost_INR", "Profit_INR"]].head()
````

## 3.4 Date features: `.dt`

| Code | Gives |
|---|---|

| `.dt.month` | 1–12 |
| `.dt.month_name()` | January… |
| `.dt.quarter` | 1–4 |
| `.dt.day_name()` | Monday… |

````python
df["Order_Date"] = pd.to_datetime(df["Order_Date"])
````

````python
df.dtypes
````

````python
df["Year"] = df["Order_Date"].dt.year
df["Month"] = df["Order_Date"].dt.month_name()
df["Quarter"] = df["Order_Date"].dt.quarter
````

````python
df[["Quantity", "Unit_Price_INR", "Gross_Sales", "Revenue_INR", "Total_Cost_INR", "Profit_INR", "Order_Date", "Year", "Month", "Quarter"]].head()
````

## 4.2 Merging on keys

A **key** is the column two tables share: `Customer_ID` links orders to customers, `Product_ID` links orders to products. Merge one table at a time.

````python
df = orders.merge(customers, on="Customer_ID", how="left")
df = df.merge(products, on="Product_ID", how="left")
````

````python
df["Gross_Sales"] = df["Quantity"]*df["Unit_Price_INR"]
df["Revenue_INR"] = df["Gross_Sales"] * (1- df["Discount_Pct"]/100)
````

## 4.3 GroupBy: split → apply → combine

```python
df.groupby("group_col")["value_col"].sum()
```

| Part | Meaning |
|---|---|
| `groupby("Region")` | Split rows into one group per region |
| `["Revenue_INR"]` | The column to calculate on |
| `.sum()` | Apply: `sum`, `mean`, `count`, `min`, `max`, `nunique` |

````python
df.groupby("Region")["Revenue_INR"].sum()
````

````python
df.groupby("Category")["Revenue_INR"].sum()
````

## 4.4 Several calculations at once: `agg()`

`agg` takes a **dict** in curly braces `{column: function}`:

```python
df.groupby("Region").agg({"Revenue_INR": "sum", "Quantity": "sum", "Order_ID": "count"})
```

Square brackets `[...]` give `SyntaxError`. Named aggregation also sets the output names:

```python
df.groupby("Region").agg(Revenue=("Revenue_INR", "sum"), Orders=("Order_ID", "count"))
```

````python
df.groupby("Region").agg({"Revenue_INR": "sum", "Quantity":"sum", "Order_ID": "count"})
````

## 4.5 Top N

Group → sum → sort descending → `head(n)`.

````python
top_five = df.groupby("Product_Name")["Revenue_INR"].sum().sort_values(ascending = False).head(5)
top_five
````

## 4.6 Stacking tables: `concat()`

| | `merge` | `concat` |
|---|---|---|
| Joins | **Side by side** on a key | **One below the other** |
| Use for | Adding columns from another table | Same columns split across files (Jan, Feb, Mar) |
| Rows | Follow the key and `how` | Sum of all rows |

`ignore_index=True` renumbers the rows 0, 1, 2… instead of repeating each file's index.

````python
jan = pd.read_csv("sales_jan.csv")
feb = pd.read_csv("sales_feb.csv")
mar = pd.read_csv("sales_mar.csv")
````

````python
all_sales = pd.concat([jan, feb, mar], ignore_index = True)
all_sales.head(1)
````

## 4.7 Validate the concat

The row counts must add up: 91 + 93 + 116 = 300. More rows means duplicates; fewer means something was lost.

````python
print(jan.shape)
print(feb.shape)
print(mar.shape)
print(all_sales.shape)
print("Rows add up:", jan.shape[0] + feb.shape[0] + mar.shape[0] == all_sales.shape[0])
````

### Selecting & filtering
| Task | Code |
|---|---|
| Columns | `df["a"]` (Series), `df[["a", "b"]]` (DataFrame) |
| By label / position | `df.loc[row, "col"]` / `df.iloc[r, c]` |
| Filter rows | `df[df["a"] > 10]` |
| Filter + columns | `df.loc[df["a"] > 10, ["a", "b"]]` |
| Combine conditions | `(c1) & (c2)`, `(c1) \| (c2)`, `~(c1)` |
| Sort | `df.sort_values(["a", "b"], ascending=[False, True])` |

### Cleaning
| Task | Code |
|---|---|
| Missing | `df.isna().sum()`, `df.dropna()`, `df["col"].fillna(value)` |
| Duplicates | `df.duplicated().sum()`, `df = df.drop_duplicates()` |
| Text | `df["col"].str.strip().str.title()` |
| Types | `df["col"].astype(int)`, `pd.to_datetime(df["col"])` |

### Transforming & combining
| Task | Code |
|---|---|
| New column | `df["new"] = df["a"] * df["b"]` |
| Date parts | `df["d"].dt.year`, `.dt.month_name()`, `.dt.quarter` |
| Merge | `a.merge(b, on="key", how="left")` |
| Stack | `pd.concat([a, b, c], ignore_index=True)` |

### Aggregating
| Task | Code |
|---|---|
| One total per group | `df.groupby("g")["v"].sum()` |
| Several | `df.groupby("g").agg({"v": "sum", "id": "count"})` |
| Top N | `df.groupby("g")["v"].sum().sort_values(ascending=False).head(5)` |

### Conversions
| Task | Code |
|---|---|
| List of dicts → DataFrame | `pd.DataFrame(records)` |
| DataFrame → list of dicts | `df.to_dict("records")` (`"records"` is a keyword, not a variable) |
