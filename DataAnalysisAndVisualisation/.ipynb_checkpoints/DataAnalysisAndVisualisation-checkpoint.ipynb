# Part 0 — Setup (MacBook Air)

## 0.1 Check which modules are installed

| Module | Needed for | Install name |
|---|---|---|
| `pandas` | DataFrames, `read_sql` | `pandas` |
| `sqlalchemy` | The engine that connects pandas to the database | `sqlalchemy` |
| `pymysql` | The MySQL **driver** (the `+pymysql` in the URL) | `pymysql` |
| `cryptography` | Needed by PyMySQL for MySQL 8's default password method | `cryptography` |

`importlib.util.find_spec(name)` checks if a module **can** be imported, without importing it. It returns `None` if the module is missing.

````python
import importlib.util
import sys

REQUIRED = {
    "pandas": "pandas",
    "sqlalchemy": "sqlalchemy",
    "pymysql": "pymysql",
    "cryptography": "cryptography",
}

print("Python used by this notebook:", sys.executable)
print()

missing = []
for module, package in REQUIRED.items():
    if importlib.util.find_spec(module) is None:
        print(f"  ✘ {module:<13} NOT installed")
        missing.append(package)
    else:
        print(f"  ✔ {module:<13} installed")

print()
print("Missing:", missing if missing else "nothing - all set!")
````

## 0.2 Install the missing modules

`sys.executable -m pip` installs into **the same Python this notebook runs on**. A plain `pip install` in Terminal can install into a different Python, which is the usual cause of `ModuleNotFoundError` even after installing.

On a Mac with Homebrew Python, pip blocks installs with `externally-managed-environment`. The flags below handle that:

| Flag | Does |
|---|---|
| `--user` | Installs into your home folder, not Homebrew's files |
| `--break-system-packages` | Allows the install on Homebrew Python (safe together with `--user`) |

After installing, **restart the kernel** (Kernel → Restart) and run 0.1 again.

````python
import subprocess

if missing:
    print("Installing:", " ".join(missing))
    subprocess.check_call([
        sys.executable, "-m", "pip", "install",
        "--user", "--break-system-packages", *missing,
    ])
    print("\nDone - now restart the kernel and run 0.1 again.")
else:
    print("Nothing to install.")
````

The same install from **Terminal**:

```bash
python3 -m pip install --user --break-system-packages pandas sqlalchemy pymysql cryptography
```

If Jupyter itself isn't installed yet: add `notebook` to that command, then start it with `python3 -m notebook`.

# Part 1 — Connecting to MySQL

## 1.1 Imports and connection settings

| Setting | Meaning | Usual value on a Mac |
|---|---|---|
| `DB_USER` | MySQL user | `root` |
| `DB_PASSWORD` | Password set in 0.3 | your password |
| `DB_HOST` | Where MySQL runs | `localhost` (your own Mac) |
| `DB_PORT` | MySQL's port | `3306` (default) |
| `DB_NAME` | Database | `world` |

**Keep passwords out of the code** when you share it: `getpass` asks for it when the cell runs. (In the class code it was typed in directly.)

````python
import pandas as pd
from sqlalchemy import create_engine, text
from getpass import getpass

DB_USER = "root"
DB_PASSWORD = getpass("MySQL password: ")   # or type it in: DB_PASSWORD = "your_password"
DB_HOST = "localhost"
DB_PORT = 3306
DB_NAME = "world"
````

## 1.2 Create the SQLAlchemy engine

```
mysql+pymysql://user:password@host:port/database
  │      │
  │      └─ driver  (the Python module that talks to MySQL)
  └──────── dialect (which database type)
```

| Term | Meaning |
|---|---|
| **SQLAlchemy** | A Python toolkit that talks to many databases the same way |
| **Engine** | SQLAlchemy's connection manager; pandas uses it to run queries |
| **Dialect + driver** | `mysql+pymysql`; others: `postgresql+psycopg2`, `sqlite:///file.db` |

`create_engine` does **not** connect yet; the first query does. A wrong password only shows up in 1.3 or 2.2.

````python
DATABASE_URL = (
    f"mysql+pymysql://{DB_USER}:{DB_PASSWORD}"
    f"@{DB_HOST}:{DB_PORT}/{DB_NAME}"
)

engine = create_engine(DATABASE_URL)
print("Engine created:", engine.url.render_as_string(hide_password=True))
````

## 1.3 Test the connection

`SELECT 1` is the smallest possible query: if it works, the connection works. `text()` wraps raw SQL for SQLAlchemy.

````python
try:
    with engine.connect() as conn:
        version = conn.execute(text("SELECT VERSION()")).scalar()
        tables = conn.execute(text("SHOW TABLES")).fetchall()
    print("✔ Connected - MySQL version:", version)
    print("Tables in", DB_NAME + ":", [t[0] for t in tables])
except Exception as e:
    print("✘ Connection failed:", e)
````

# Part 2 — Reading SQL into pandas

## 2.1 The SQL query

All cities in **India** in the district **Tamil Nadu**. Triple quotes `"""..."""` let the query span several lines.

| SQL | pandas equivalent |
|---|---|
| `SELECT *` | all columns |
| `FROM city` | the table |
| `WHERE CountryCode = 'IND'` | `df[df["CountryCode"] == "IND"]` |
| `AND District = 'Tamil Nadu'` | `& (df["District"] == "Tamil Nadu")` |

Filtering **in SQL** is better than loading everything and filtering in pandas: only the rows you need travel to Python.

````python
query = """
SELECT *
FROM city
WHERE CountryCode = 'IND'
  AND District = 'Tamil Nadu'
"""
````

## 2.2 Read MySQL data into pandas: `pd.read_sql()`

`pd.read_sql(query, engine)` runs the query and returns a **DataFrame**; column names come from the table.

`try` / `except` catches connection problems and prints the error instead of crashing.

In the class `.py` script, `exit()` stopped the script on an error. It's left out here, because in Jupyter `exit()` shuts down the kernel.

````python
try:
    df = pd.read_sql(query, engine)

    print("\nData loaded successfully!")
    print("-" * 60)

except Exception as e:
    print("Database error:", e)
````

| Reader | Use |
|---|---|
| `pd.read_sql(query, engine)` | Any SQL query |
| `pd.read_sql_table("city", engine)` | A whole table, no SQL |
| `pd.read_sql(query, engine, params={...})` | Query with values passed safely (see 5.2) |
| `df.to_sql("name", engine, if_exists="replace", index=False)` | Write a DataFrame **back** to MySQL |

# Part 4 — Filtering & Sorting

## 4.1 One condition

```python
df[df["Population"] > 500000]
```

Write numbers **without commas**: `500000`, or `500_000` for readability. `500,000` would be a tuple.

````python
print("\n--- CITIES WITH POPULATION > 500,000 ---")

big = df[df["Population"] > 500000]
big[["Name", "District", "Population"]]
````

## 4.2 Two conditions (class version)

| Operator | Meaning |
|---|---|
| `&` | AND |
| `\|` | OR |
| `~` | NOT |

Put **each condition in its own brackets**. Without them, `500000 & df["Population"]` runs first and raises an error.

````python
print("\n--- CITIES WITH POPULATION BETWEEN 100,000 AND 500,000 ---")

large_cities = df[
    (df["Population"] < 500000) &
    (df["Population"] > 100000)
]

large_cities[["Name", "District", "Population"]]
````

### Same filter, other ways

| Way | Code |
|---|---|
| `between` | `df[df["Population"].between(100000, 500000, inclusive="neither")]` |
| `loc` (rows + columns at once) | `df.loc[cond, ["Name", "District", "Population"]]` |
| In SQL | `WHERE Population > 100000 AND Population < 500000` |

`inclusive="neither"` keeps the strict `<` and `>`; the default `"both"` would include 100,000 and 500,000.

````python
same = df.loc[
    df["Population"].between(100000, 500000, inclusive="neither"),
    ["Name", "District", "Population"],
]
print("Same rows as the class version:", same.equals(large_cities[["Name", "District", "Population"]]))
````

## 4.3 Sorting: `sort_values()`

| Option | Does |
|---|---|
| `by="Population"` | Column to sort on |
| `ascending=False` | Biggest first |
| `by=["a", "b"]` | Sort by `a`, ties by `b` |
| `kind="stable"` | Ties keep their original order |

````python
print("\n--- CITIES SORTED BY POPULATION ---")

sorted_cities = df.sort_values(
    by="Population",
    ascending=False
)

sorted_cities[["Name", "District", "Population"]]
````

# Part 5 — Notes

## 5.1 Common mistakes and errors

| Error / mistake | Cause | Fix |
|---|---|---|
| `ModuleNotFoundError: No module named 'pymysql'` | Driver not installed, or installed into another Python | Run 0.1 / 0.2 (installs with `sys.executable`) |
| `RuntimeError: 'cryptography' package is required` | MySQL 8's password method needs it | Install `cryptography` |
| `Can't connect to MySQL server on 'localhost'` | MySQL isn't running | `brew services start mysql` |
| `Access denied for user 'root'@'localhost'` | Wrong user or password | Check `DB_USER` / `DB_PASSWORD` |
| `Unknown database 'world'` | Sample database not loaded | Section 0.4 |
| `Table 'world.City' doesn't exist` | Table names are case-sensitive on Mac/Linux | Use `city` |
| `externally-managed-environment` | Homebrew Python blocks pip | `--user --break-system-packages` |
| Extra `None` after `info()` | `print(df.info())` | Just `df.info()` |
| `TypeError` in a two-condition filter | Missing brackets | `(a) & (b)` |
| `500,000` in code | Commas make a tuple | `500000` or `500_000` |
| Password in shared code | Hard-coded secret | `getpass()` or an environment variable |

## 5.2 SQL injection (good to know)

Never build a query by gluing in user input:

```python
district = input()
query = f"SELECT * FROM city WHERE District = '{district}'"     # ✘ unsafe
```

Pass values as **parameters** instead; the driver escapes them:

```python
query = text("SELECT * FROM city WHERE District = :d")
df = pd.read_sql(query, engine, params={"d": "Tamil Nadu"})       # ✔ safe
```

## 5.3 Interview points

| Question | Short answer |
|---|---|
| What is SQLAlchemy's engine? | A connection manager pandas uses to run SQL against a database |
| What's in a connection URL? | `dialect+driver://user:password@host:port/database` |
| Why a driver (PyMySQL)? | SQLAlchemy needs a Python module that speaks MySQL's protocol |
| `read_sql` vs `read_csv`? | `read_sql` runs a query on a live database; `read_csv` reads a file |
| Filter in SQL or pandas? | In SQL when possible: less data moves to Python |
| How do you check missing values? | `df.isnull().sum()` |
| Why brackets in `(a) & (b)`? | `&` runs before `<` / `>`, so each comparison needs brackets |
| How do you write back to the database? | `df.to_sql("table", engine, if_exists="replace", index=False)` |
| How do you avoid SQL injection? | Query parameters (`params=`), never f-strings with user input |
